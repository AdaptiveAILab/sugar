```python
import torch
import numpy as np


def to_np(x: torch.Tensor) -> np.ndarray:
    return x.detach().cpu().numpy()

# Generate inputs
x = torch.linspace(-5, 5, 1000, requires_grad = True)

# ----------------------------------------------------------------
y = torch.relu(x)
# ----------------------------------------------------------------

# Backprop
y.sum().backward()
dydx = x.grad

import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots
from matplotlib.ticker import MaxNLocator

mpl.rcdefaults()

plt.style.use(["science", "nature"])


fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(1.7, 1.8), sharex=True,
    gridspec_kw=dict(hspace=0.15)
)



# ReLU
ax1.plot(to_np(x), to_np(y), color="black", label="Forward")
#ax1.set_ylabel("y")

ax1.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax1.set_ylim(-0.5, 5.0)


ax1.set_title("Plain ReLU", loc="center", pad=6)

# Gradient
ax2.plot(to_np(x), to_np(dydx), color="#C83200", label="Backward")
#ax2.set_ylabel("grad")
ax2.set_xlabel("$x$")

ax2.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax2.set_ylim(-0.2, 1.4)

for ax in (ax1, ax2):
    ax.xaxis.set_major_locator(MaxNLocator(4))
    ax.yaxis.set_major_locator(MaxNLocator(3))
    ax.minorticks_off()
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(top=False, right=False)
    ax.grid(False)


fig.tight_layout(pad=0.25)
fig.savefig("relu_grad_stack.pdf", bbox_inches="tight", pad_inches=0.01)
```

```python
def to_np(x: torch.Tensor) -> np.ndarray:
    return x.detach().cpu().numpy()

# Generate inputs
x = torch.linspace(-5, 5, 1000, requires_grad = True)

def nelu(x, alpha=0.2):
    return torch.where(
        x > 0,
        x,
        alpha * (x * x) / (1 + x * x)
    )

m = nelu(x)
y = m - m.detach() + torch.relu(x).detach()
# ----------------------------------------------------------------

# Backprop
y.sum().backward()
dydx = x.grad

import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots
from matplotlib.ticker import MaxNLocator

mpl.rcdefaults()

plt.style.use(["science", "nature"])


fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(1.7, 1.8), sharex=True,
    gridspec_kw=dict(hspace=0.15)
)


# ReLU
ax1.plot(to_np(x), to_np(y), color="black", label="Forward")
#ax1.set_ylabel("y")

ax1.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax1.set_ylim(-0.5, 5.0)


ax1.set_title("SUGAR(NeLU)", loc="center", pad=6)

# Gradient
ax2.plot(to_np(x), to_np(dydx), color="#C83200", label="Backward")
#ax2.set_ylabel("grad")
ax2.set_xlabel("$x$")
ax2.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax2.set_ylim(-0.2, 1.4)


for ax in (ax1, ax2):
    ax.xaxis.set_major_locator(MaxNLocator(4))
    ax.yaxis.set_major_locator(MaxNLocator(3))
    ax.set_yticks([])
    ax.set_yticks([])
    ax.set_ylabel(None)
    ax.set_ylabel(None)
    ax.minorticks_off()
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(top=False, right=False)
    ax.grid(False)



fig.tight_layout(pad=0.25)
fig.savefig("sugar_grad_stack.pdf", bbox_inches="tight", pad_inches=0.01)
```

```python
def to_np(x: torch.Tensor) -> np.ndarray:
    return x.detach().cpu().numpy()

# Generate inputs
x = torch.linspace(-5, 5, 1000, requires_grad = True)

def b_silu(x: torch.Tensor, alpha: float = 1.67) -> torch.Tensor:
    return (x + alpha) * torch.sigmoid(x) - alpha / 2

m = b_silu(x)
y = m - m.detach() + torch.relu(x).detach()
# ----------------------------------------------------------------

# Backprop
y.sum().backward()
dydx = x.grad

import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots
from matplotlib.ticker import MaxNLocator

mpl.rcdefaults()

plt.style.use(["science", "nature"])


fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(1.7, 1.8), sharex=True,
    gridspec_kw=dict(hspace=0.15)
)


# ReLU
ax1.plot(to_np(x), to_np(y), color="black", label="Forward")
#ax1.set_ylabel("y")

ax1.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax1.set_ylim(-0.5, 5.0)

ax1.set_title("SUGAR(B-SiLU)", loc="center", pad=6)

# Gradient
ax2.plot(to_np(x), to_np(dydx), color="#C83200", label="Backward")
#ax2.set_ylabel("grad")
ax2.set_xlabel("$x$")
ax2.legend(
    frameon=False,
    loc="upper left",
    labelcolor="linecolor",
    handlelength=0,
    handletextpad=0,
)

ax2.set_ylim(-0.2, 1.4)


for ax in (ax1, ax2):
    ax.xaxis.set_major_locator(MaxNLocator(4))
    ax.yaxis.set_major_locator(MaxNLocator(3))
    ax.minorticks_off()
    ax.set_yticks([])
    ax.set_yticks([])
    ax.set_ylabel(None)
    ax.set_ylabel(None)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(top=False, right=False)
    ax.grid(False)



fig.tight_layout(pad=0.25)
fig.savefig("sugar_bsilu_grad_stack.pdf", bbox_inches="tight", pad_inches=0.01)
```
